```python
# Simple Calculator
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print(f'First no is {num1} and Second no is {num2}')

print("Addition is simple: ", num1 + num2)
print("Subtraction is simple: ", num1 - num2)
print("Multiplication is simple: ", num1 * num2)
print("Division is simple: ", num1 / num2)

# Temperature Converter
c = float(input("Enter the temperature in Celsius: "))
f = float((c * 1.8) + 32)

print("Your temperature in Fahrenheit is", f)
```
